## ¿Qué es una serie de tiempo?

Una serie de tiempo o serie temporal es una secuencia de datos, observaciones o valores, medidos en determinados momentos y ordenados cronológicamente. Los datos pueden estar espaciados a intervalos iguales o desiguales. Una vez que se captura una serie de tiempo, a menudo se realiza un análisis sobre ella para identificar patrones en los datos, en esencia, lo que se busca es entender que suceda a medida que el tiempo va avanzando. Ser capaz de procesar datos de series de tiempo es una habilidad esencial en el mundo moderno. Uno de los usos más habituales de las series de tiempo es su análisis para predicción y pronóstico (así se hace por ejemplo con los datos climáticos, las acciones de bolsa, o las series de datos demográficos). Resulta difícil imaginar una rama de las ciencias en la que no aparezcan datos que puedan ser considerados como series de tiempo.

### ¿Qué las hace especiales?

Las características que hacen a las series de tiempo especiales y las diferencia de, por ejemplo, un problema de regresión son las siguientes:

1. **Son dependientes del tiempo**; por lo tanto el supuesto básico de los modelos de regresión de que las observaciones son independientes no se sostiene en este caso.

2. **Suelen tener una tendencia**; la mayoría de las series de tiempo suelen tener algún tipo de tendencia de estacionalidad, es decir, las variaciones propias de un período de tiempo determinado.

3. **Suelen estar autocorrelacionadas**; las observaciones consecutivas estén correlacionadas. Esta correlación entre observaciones consecutivas se llama autocorrelación. Cuando los datos están autocorrelacionados, la mayoría de los métodos estadísticos estándares basados en la suposición de observaciones independientes pueden arrojar resultados engañosos o incluso ser inútiles.

## Series de tiempo con Python

Las principales librerías que nos ofrece [Python](https://python.org/) para trabajar con [series de tiempo](https://es.wikipedia.org/wiki/Serie_temporal) son:

* **[Statsmodels](https://statsmodels.sourceforge.net/stable/tsa.html)**: Esta librería contiene muchos objetos y funciones de suma utilidad para el análisis de series de tiempo. Algunos de los modelos que están cubiertos por Statsmodels incluyen: el modelo autorregresivo (AR); el modelo autorregresivo de vectores (VAR); y el modelo autorregresivo de media móvil (ARMA). También incluye funciones de estadística descriptiva de series de tiempo, como por ejemplo la autocorrelación, así como las correspondientes propiedades teóricas de ARMA o procesos relacionados. Por último, también ofrece las pruebas estadísticas relacionadas y algunas funciones auxiliares muy útiles.


* **[Pandas](https://pandas.pydata.org/)**: Pandas proporciona un amplio soporte para trabajar con datos de series de tiempo. Generalmente cuando trabajamos con series de tiempo realizamos un amplio abanico de tareas, como: convertir fechas, estandarizar el tiempo de acuerdo a la zona horaria, crear secuencias a determinados intervalos o frecuencias, identificar datos faltantes, desplazar las fechas hacia atrás o hacia adelante por un determinado valor, calcular resúmenes agregados de valores a medida que el tiempo cambia, etc. Pandas nos brinda las herramientas para poder realizar estas y muchas otras tareas en forma muy sencilla.


**Importamos los módulos necesarios**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib as mpl
plt.rcParams.update({'figure.figsize': (10, 7), 'figure.dpi': 120})
```

**Leemos nuestra serie de tiempo desde un archvo csv**

```python
# Importar datos en formato csv.
co2 = pd.read_csv('data/co2.csv',encoding='latin-1', sep = ';')
```

```python
#Validamos la correcta carga de la base de datos
#===============================================
co2.head(5)
```

```python
# Generamos la variable fecha para formar nuestra serie de tiempo
co2['fecha'] = co2.Year.astype(str) + str('-') + co2.Month.astype(str) + str('-1')
co2.head(5)
```

```python
co2.shape
```

```python
# Validmos el correcto tipo de las variables
# co2.dtypes
co2.fecha = pd.to_datetime(co2.fecha)
```

```python
# validamos los cambios de variables
# co2.dtypes
co2.head(10)
```

```python
co2.dtypes
```

**Graficamos la serie de tiempo**

```python
plt.figure(figsize=(16,5), dpi=100)
plt.plot(co2.fecha, co2.CO2,color='tab:blue')
plt.xlabel('Periodos')
plt.ylabel('CO2')
plt.title('Datos CO2', fontsize=16)
#plt.gca().set(title='Serie CQI', xlabel=df_1.RESULT_TIME, ylabel=df_1.CQI)
plt.show()
```

```python
sns.boxplot(x = 'Month', y = 'CO2', data = co2)
```

**Graficamos las cajas agregadas**

```python
# Draw Plot
fig, axes = plt.subplots(1, 2, figsize=(20,7), dpi= 80)
sns.boxplot(x='Year', y='CO2', data=co2, ax=axes[0])
sns.boxplot(x='Month', y='CO2', data=co2, ax=axes[1])
```

**Agrupando periodos**

```python
# agrupando por periodos específicos
co2_mes = co2
```

```python
# co2_mes.reset_index(inplace = True)
co2_mes.set_index('fecha',inplace = True)
```

```python
co2_mes.head(10)
```

```python
# agrupamos los datos
co2_mes = co2_mes.resample('M').sum()
co2_year = co2_mes.resample('Y').mean()
co2_quarter = co2_mes.resample('Q').mean()
```

```python
print(co2_mes.shape)
print(co2_year.shape)
print(co2_quarter.shape)
```

```python
co2_quarter.head(14)
```

```python
plt.figure(figsize=(16,5), dpi=100)
plt.plot(co2.CO2,color='tab:green',label='Mensual')
plt.plot(co2_year.CO2,color='tab:blue',label='Anual')
plt.plot(co2_quarter.CO2,color='tab:red',label='Trimestral')
#plt.plot(co2_mes.CO2,color='tab:brown')
plt.legend()
plt.show()
```
